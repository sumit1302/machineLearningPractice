```python

import pandas as pd
import numpy as np
import matplotlib
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_squared_error, r2_score
```

```python
df = pd.read_csv('USvideos.csv')
```

```python
df.head()
```

```python
df.tail()
```

```python
df.columns
```

```python
x= df[['likes','comment_count']]
y= df[['views']]
```

```python

# split the dataset
X_train, X_test, y_train, y_test = train_test_split(
    x, y, test_size=0.05, random_state=0)
```

```python
X_train
```

```python
X_test
```

```python
y_train
```

```python
# Using model Linear Regression
model = LinearRegression()
```

```python
#training the data
model.fit(X_train,y_train)
```

```python
#predicting the views for test data
views_pred=model.predict(X_test)
```

```python
#checking the model
meanSquare = mean_squared_error(y_test, views_pred)
r2= r2_score(y_test,views_pred)

print(f"Mean squared error : {meanSquare}")
print(f"R-squared : {r2}")
```
